#Convert Date Columns

```python
date_columns = ['booking_date', 'check_in_date', 'check_out_date',
                'actual_check_in', 'actual_check_out']

for col in date_columns:
    reservations[col] = pd.to_datetime(reservations[col], errors='coerce')
```

```python
reservations[date_columns].dtypes
```

#Which property generates the highest revenue?

```python
reservations[reservations['booking_status'] == 'Checked Out'].merge(properties, on='property_id').groupby('property_name')['booking_value'].sum().sort_values(ascending=False)
```

#Which booking channel generates the highest revenue?

```python
reservations[reservations['booking_status'] == 'Checked Out'].groupby('booking_channel')['booking_value'].sum().sort_values(ascending=False)
```

#Which room type generates the highest revenue?

```python
reservations.merge(rooms, on='room_id').query("booking_status == 'Checked Out'").groupby('room_type')['booking_value'].sum().sort_values(ascending=False)
```

#Which guest type contributes the highest revenue?

```python
reservations[reservations['booking_status'] == 'Checked Out'].merge(guests, on='guest_id').groupby('guest_type')['booking_value'].sum().sort_values(ascending=False)
```

#Who are the top 5 guests by revenue?

```python
reservations[reservations['booking_status'] == 'Checked Out'].groupby('guest_id')['booking_value'].sum().nlargest(5)
```

#What is the average length of stay by room type?

```python
reservations.merge(rooms, on='room_id').query("booking_status == 'Checked Out'").groupby('room_type')['nights_booked'].mean().round(2)
```

#Do loyalty members generate more revenue?

```python
reservations[reservations['booking_status'] == 'Checked Out'].merge(guests, on='guest_id').groupby('loyalty_member')['booking_value'].sum()
```

#Which guest cities generate the highest revenue?

```python
reservations[reservations['booking_status'] == 'Checked Out'].merge(guests, on='guest_id').groupby('home_city')['booking_value'].sum().nlargest(5)
```

#Find the highest and lowest booking values

```python
print("Highest:", reservations['booking_value'].max())
print("Lowest:", reservations['booking_value'].min())
```

#Visualize booking status distribution

```python
reservations['booking_status'].value_counts().plot(kind='bar')
plt.title('Booking Status Distribution')
plt.xlabel('Booking Status')
plt.ylabel('Number of Bookings')
plt.show()
```

#Compare revenue by property

```python
data = reservations[reservations['booking_status'] == 'Checked Out'].merge(
    properties, on='property_id'
)

data.groupby('property_name')['booking_value'].sum().plot(kind='bar')
plt.title('Revenue by Property')
plt.ylabel('Revenue')
plt.show()
```
